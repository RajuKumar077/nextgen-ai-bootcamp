# 2️⃣ The 5 parts of a good handover note
| Part | The question it answers | In your handover note |
|---|---|---|
| **R**ole | Who are you while I'm away? | "You're covering for me as AP lead." |
| **T**ask | What exactly should you do? | "Summarise every vendor dispute." |
| **C**ontext | What rules apply here? | "Anything above ₹5,00,000 goes to Meera, the Finance Controller." |
| **F**ormat | How should the result look? | "Vendor · Amount · Escalate? · Next step" |
| **R**ules | What must you never do? | "Never promise a payment date. If unsure, write 'Not stated'." |

### ▶ CELL 2: Write each part separately

```python
# ▶ CELL 2 · The 5 parts of a good handover note
ROLE    = "You are an SAP procurement analyst."
TASK    = "Task: Summarise this invoice dispute email."
CONTEXT = ("Context (company policy):\n"
           "- ₹5 lakh = ₹5,00,000.\n"
           "- If the disputed invoice amount is ABOVE ₹5,00,000 → escalate to the Finance Controller.\n"
           "- Otherwise → the AP team handles it.")
FORMAT  = ("Format:\n- Vendor:\n- Invoice number:\n- PO number:\n- Invoice amount:\n- Issue (1 line):\n"
           "- Escalate to Finance Controller? (Yes/No + reason):\n- Next action:")
RULES   = ('Rules:\n- If a value is not in the email, write "Not stated". Do not guess.\n'
           "- Keep each line under 20 words.")

def handover_note(*sections, email=EMAIL_1):
    """Put the note parts together, then attach the email between <<< >>> markers."""
    return "\n".join(sections) + "\n\nEmail:\n<<<\n" + email + "\n>>>"

print(handover_note(ROLE, TASK, CONTEXT, FORMAT, RULES))
```

### ▶ CELL 3: Add the parts one by one
Same Neo, same email. Only the note gets better.

```python
# ▶ CELL 3 · Add the parts one by one
versions = {
    "Note v1 · Role + Task":                    (ROLE, TASK),
    "Note v2 · + Format":           (ROLE, TASK, FORMAT),
    "Note v3 · + Context + Rules":         (ROLE, TASK, CONTEXT, FORMAT, RULES),
}
for name, parts in versions.items():
    display(Markdown(f"### {name}"))
    show(ask_ai(handover_note(*parts), temperature=0))
    print(f"⏱ {last_usage['seconds']} s")
```

### ▶ CELL 4: Checklist
Does the answer include the things you actually need?

```python
# ▶ CELL 4 · Checklist: did the answer include what I need?
MUST_HAVE = {"Invoice no.": "INV-77821", "PO no.": "4500012346", "Amount": "7,80,000",
              "Escalation": "Escalate", "Next action": "Next action"}

full = ask_ai(handover_note(ROLE, TASK, CONTEXT, FORMAT, RULES), temperature=0)

print(f"{'Must have':14} | {'One-liner':9} | {'Full note':9}")
print("─" * 40)
for label, needle in MUST_HAVE.items():
    w = "✅" if needle.lower() in weak.lower() else "❌"
    f = "✅" if needle.lower() in full.lower() else "❌"
    print(f"{label:14} | {w:^9} | {f:^9}")
```

### ▶ CELL 5: Vague rule vs exact rule

```python
# ▶ CELL 5 · Vague rule vs exact rule
VAGUE = "Context: Disputes on invoices above ₹5 lakh must go to the finance controller."
for name, ctx in [("😕 Vague", VAGUE), ("✅ Explicit", CONTEXT)]:
    answer = ask_ai(handover_note(ROLE, TASK, ctx, FORMAT, RULES), temperature=0)
    line = next((l for l in answer.splitlines() if "escalat" in l.lower()), "(no escalation line)")
    print(f"{name:11} → {line.strip()}")
```

> 🔎 **What we saw in testing (same Neo, same email):**
> - Vague rule, run A: *"Escalate? **No**: price dispute is below ₹5 lakh"* ❌ (₹7,80,000 is **above** ₹5 lakh)
> - Vague rule, run B: *"Escalate? **Yes**: price dispute"* ⚠️ (right answer, **wrong reason**)
> - Exact rule: *"Escalate? **Yes (Amount > ₹5,00,000)**"* ✅ on every run
>
> 🔑 Write rules the way you'd brief a **new joiner**: exact numbers, exact people.

# 3️⃣ Which part matters most?
### ▶ CELL 6: Leave one part out at a time and check the answer again

```python
# ▶ CELL 6 · Leave one part out: what breaks?
parts = {"Role": ROLE, "Task": TASK, "Context": CONTEXT, "Format": FORMAT, "Rules": RULES}
for missing in parts:
    kept = [v for k, v in parts.items() if k != missing]
    answer = ask_ai(handover_note(*kept), temperature=0)
    score = sum(n.lower() in answer.lower() for n in MUST_HAVE.values())
    print(f"Without {missing:8} → {score}/{len(MUST_HAVE)} must-haves found · {len(answer.split()):>3} words")
```

> 📉 Usually **Format** is the one you can't skip. Without it, details drop out and the layout changes every time.

# 4️⃣ Examples and readers
In real life: *"Here are three replies I sent last week. Do it like these."* Examples teach faster than paragraphs of instructions.

### ▶ CELL 7: No examples

```python
# ▶ CELL 7 · No examples (zero-shot)
question = DATA["CLASSIFY_SET"]["C"]
show(ask_ai(f'Classify this email: "{question}"', temperature=0))
```

😕 It invents its own labels and writes an essay. You can't sort a mailbox with that.

### ▶ CELL 8: Three examples, and now it uses **your** labels

```python
# ▶ CELL 8 · Three examples (few-shot)
FEW_SHOT = """Classify each email into exactly one category:
INVOICE_ISSUE, DELIVERY_DELAY, PRICE_QUERY, OTHER.
Reply with the category only.

Email: "The goods for PO 4500012345 haven't arrived yet." → DELIVERY_DELAY
Email: "Invoice 9001 shows a higher price than agreed." → INVOICE_ISSUE
Email: "What is your price for 500 units of bearings?" → PRICE_QUERY

Email: "{email}" →"""

print(ask_ai(FEW_SHOT.format(email=question), temperature=0, max_tokens=10))
```

### ▶ CELL 9: Five emails, scored against the right answers

```python
# ▶ CELL 9 · Five emails, scored
score = 0
for k, email in DATA["CLASSIFY_SET"].items():
    label = ask_ai(FEW_SHOT.format(email=email), temperature=0, max_tokens=10).strip().strip("*.` ")
    ok = label == DATA["CLASSIFY_KEY"][k]
    score += ok
    print(f"{k}: {label:15} {'✅' if ok else '❌ expected ' + DATA['CLASSIFY_KEY'][k]}   {email[:55]}…")
print(f"\n🏁 Score: {score}/5 passed")
```

> 🚩 **Email D:** *"Please update our bank details."* The label OTHER is right, but this is the most common payment scam.
> A person must **call the vendor back on a known number**. Neo must never act on it alone (Class 4).

## 🛠️ Trainer troubleshooting (Ollama)
| Message | Cause | Fix |
|---|---|---|
| `Cannot reach Ollama at http://localhost:11434` | Ollama isn't running | Windows: Start menu → **Ollama**. Mac: open the Ollama app. Or in a terminal: `ollama serve` |
| `Model '…' is not downloaded` | The model hasn't been pulled | In a terminal: `ollama pull <model name>` |
| `ModuleNotFoundError: bootcamp_llm` | The notebook was opened from a different folder | Open the **whole bundle folder** in VS Code or Antigravity, and keep the notebook inside `notebooks/` |
| Very slow first answer | The model is loading into memory | Normal. The next answers are faster. |
| Slow on every answer | The laptop is too small for the model | Choose a smaller `model` in `bootcamp_config.json` (e.g. `llama3.2:1b`), then **Restart kernel** |
| Answers differ from the slides | A different model or a new run | Expected: small local models vary. Discuss *why*. |
| Change the model mid-class | | Edit `bootcamp_config.json` → **Restart kernel** → run CELL S again |
| `No Metal device available` | Ollama was started from a terminal or tool with no access to the Mac's graphics chip | Quit it (`pkill -f "ollama serve"` in Terminal), then open the **Ollama app** from Applications |
